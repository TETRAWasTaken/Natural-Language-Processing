```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

def preprocess_review(text):
    text_lower = text.lower()
    text_no_punct = re.sub(r'[^\w\s]', '', text_lower)

    text_no_url = re.sub(r'https?://\S+|www\.\S+', '', text_no_punct)
    text_clean = re.sub(r'\d+', '', text_no_url).strip()

    tokens = word_tokenize(text_clean)
    stop_words = set(stopwords.words('english'))
    filtered_tokens = [word for word in tokens if word not in stop_words]

    return filtered_tokens

sample_input = "I love this product! It's amazing and works perfectly. Check it out at https://example.com. 5 stars!"
processed = preprocess_review(sample_input)
print(processed)
```

```python
import re 
import nltk 
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

def clean_words(text):
    text_no_html = re.sub(r'<[^>]+>', '', text)
    text_no_special = re.sub(r'[^a-zA-Z0-9\s]', '', text_no_html)

    text_no_nums = re.sub(r'\b\d+\b', '', text_no_special)
    text_lower = text_no_nums.lower()

    tokens = word_tokenize(text_lower)
    stemmer = PorterStemmer()
    stemmed_tokens = [stemmer.stem(token) for token in tokens]

    return stemmed_tokens

sample_input = "I love this product! It's amazing and works perfectly. Check it out at https://example.com. 5 stars!"
processed = clean_words(sample_input)

print(processed)
```

```python
import re 
import nltk 
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

CONTRACTIONS = {
    "don't" : "do not",
    "can't" : "cannot",
    "won't" : "will not",
    "i'm" : "i am",
    "it's" : "it is",
    "you're" : "you are",
    "they're" : "they are",
    "isn't" : "is not",
    "aren't" : "are not",
}

def preprocess_text(text):
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'#\w+', '', text)
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    words = text.split()
    expanded_words = [CONTRACTIONS.get(word.lower(), word) for word in words]
    text = ' '.join(expanded_words)

    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)

    tokens = word_tokenize(text)
    stop_words = set(stopwords.words('english'))
    filtered_tokens = [word for word in tokens if word not in stop_words]

    return filtered_tokens

sample_input = "I love this product! It's amazing and works perfectly. Check it out at https://example.com. 5 stars!"
processed = preprocess_text(sample_input)

print(processed)
```

```python
import re 
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

def preprocess_review(text):
    text_clean = re.sub(r'[^\w\s]', '', text.lower())
    tokens = word_tokenize(text_clean)

    stop_words = set(stopwords.words('english'))
    filtered_tokens = [word for word in tokens if word not in stop_words]

    lemmatizer = WordNetLemmatizer()
    lemmatized_tokens = [lemmatizer.lemmatize(token) for token in filtered_tokens]

    return filtered_tokens, lemmatized_tokens

sample_input = "I love this product! It's amazing and works perfectly. Check it out at https://example.com. 5 stars!"
before_lemmatization, after_lemmatization = preprocess_review(sample_input)

print("Before Lemmatization:", before_lemmatization)
print("After Lemmatization:", after_lemmatization)
```

```python
import re 
import nltk 
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

CONTRACTIONS = {
    "don't" : "do not",
    "can't" : "cannot",
    "won't" : "will not",
    "i'm" : "i am",
    "it's" : "it is",
    "you're" : "you are",
    "they're" : "they are",
    "isn't" : "is not",
    "aren't" : "are not",
}

def normalise_text(text):
    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    text = re.sub(r'(.)\1{2,}', r'\1', text)

    words = text.split()
    words = [CONTRACTIONS.get(word, word) for word in words]
    text = ' '.join(words)

    text = re.sub(r'[^\w\s]', '', text)
    tokens = word_tokenize(text)

    negations = {"not", "no", "never", "nor", "neither", "cannot"}
    custom_stopwords = set(stopwords.words('english')) - negations

    filtered_tokens = [word for word in tokens if word not in custom_stopwords]
    return filtered_tokens

sample_input = "I love this product! It's amazing and works perfectly. Check it out at https://example.com. 5 stars!"
processed = normalise_text(sample_input)

print(processed)
```

```python
import re 
import pandas as pd 
import nltk 
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

class NLPPipeline:
    def __init__(self):
        self.stop_words = set(stopwords.words('english'))
        self.lemmatizer = WordNetLemmatizer()

    def clean_record(self, text):
        if not isinstance(text, str) or not text.strip():
            return None

        text = re.sub(r'<[^>]+>', '', text)
        text = re.sub(r'https?://\S+|www\.\S+', '', text)
        text = re.sub(r'[^\w\s]', '', text)
        text = re.sub(r'\s+', ' ', text).strip()

        tokens = word_tokenize(text.lower())

        clean_tokens = [
            self.lemmatizer.lemmatize(w)
            for w in tokens if w not in self.stop_words
        ]

        return " ".join(clean_tokens) if clean_tokens else None

    def process_dataset(self, text_list):
        df = pd.DataFrame({'original_text': text_list})

        df['processed_text'] = df['original_text'].apply(self.clean_record)
        df = df.dropna(subset=['processed_text'])

        df = df.drop_duplicates(subset=['processed_text']).reset_index(drop=True)

        return df

sample_dataset = [
    "<p>The students are learning NLP!</p>",
    "NLP is amazing! Check out https://example.com for more info.",
    "",
    None
]

pipeline = NLPPipeline()
result_df = pipeline.process_dataset(sample_dataset)

print(result_df)
```
