```python
from Scweet import Scweet
import pandas as pd
import nest_asyncio
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
nest_asyncio.apply()
```

```python
AUTH_TOKEN = "0256bccb3f099ba5300bfec6f021163053c88ed8"
```

```python
search = "SpaceX"
since = "2026-08-01"
till = "2026-08-31"
max_tweets = 10
```

```python
scweet = Scweet(auth_token=AUTH_TOKEN)
```

```python
tweets = scweet.search(
    search,
    since=since,
    until=till,
    limit=max_tweets,
    save=True,
    save_format="csv",
    save_name="my_tweets",
)

print(f"Saved {len(tweets)} tweets to my_tweets.csv")
```

```python
data = pd.read_csv("outputs/my_tweets.csv")
display(data)
```

```python
import re
def clean_tweet(tweet):
    tweet = re.sub(r"http\S+", "", tweet)  # Remove URLs
    tweet = re.sub(r"@\S+", "", tweet)     # Remove mentions
    tweet = re.sub(r"#\S+", "", tweet)     # Remove hashtags
    tweet = re.sub(r"\s+", " ", tweet)     # Remove extra whitespace
    tweet = re.sub(r"[^\w\s]", "", tweet)     # Remove punctuation
    return tweet.strip()
```

```python
data["clean_tweet"] = data["text"].apply(clean_tweet)
display(data[["text", "clean_tweet"]])
```

```python
confidence = []
annotation = []
for text in data["clean_tweet"]:
    scores = SentimentIntensityAnalyzer().polarity_scores(text)
    compound_score = scores["compound"]

    if compound_score >= 0.05:
        label = "Positive"
    elif compound_score <= -0.05:
        label = "Negative"
    else:
        label = "Neutral"

    confidence.append(compound_score)
    annotation.append(label)
```

```python
data["compound_score"] = confidence
data["label"] = annotation
display(data[["clean_tweet", "compound_score", "label"]])
```
