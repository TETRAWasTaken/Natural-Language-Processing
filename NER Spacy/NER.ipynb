# NER Spacy

```python
import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp("Apple is looking at buying U.K. startup for $1 billion")
```

```python
for token in doc:
    print(token.text, token.pos_, token.dep_)
```

```python
ner = []
ner.append(["Text", "Label"])
for ent in doc.ents:
    ner.append([ent.text, ent.label_])
```

```python
from tabulate import tabulate
print(tabulate(ner, headers="firstrow", tablefmt="fancy_grid"))
```

```python
from spacy import displacy
displacy.serve(doc, style="ent")
```

```python
displacy.render(doc, style="ent", jupyter=True)
```

# Regular Expression ( REGEX )

```python
import re
```

```python
text = "The system migration scheduled for this weekend requires immediate coordination across several departmental teams. If you experience authentication errors or login failures, please report the issue to helpdesk@global-core.net or call the technical support line at +1-555-014-9281 for immediate troubleshooting. For questions regarding database schema updates and API integrations, contact the engineering team at dev.support@nexustech-labs.org or reach their on-call specialist at +1-555-017-4432. Any compliance, data privacy, or security policy concerns must be forwarded directly to compliance-officer@secure-data.io, or you can leave a secure message on their dedicated hotline at +1-800-555-0199."
```

```python
regex_email = r"[a-z0-9.]+@[a-z0-9-.]+"
emails = re.findall(regex_email, text)
for i in emails:
    print(i)
```

```python
phone_regex = r"\+1-\d{3}-\d{3}-\d{4}"
phone_numbers = re.findall(phone_regex, text)
for number in phone_numbers:
    print(number)
```

# Emoji Handling

```python
import emoji
```

```python
emoji_text = """
🚀 The system migration scheduled for this weekend requires immediate coordination across several departmental teams. 🛠️ If you experience authentication errors or login failures, please report the issue to helpdesk@global-core.net 📧 or call the technical support line at +1-555-014-9281 📞 for immediate troubleshooting. 💻 For questions regarding database schema updates and API integrations, contact the engineering team at dev.support@nexustech-labs.org 📩 or reach their on-call specialist at +1-555-017-4432. 📱 Any compliance, data privacy, or security policy concerns must be forwarded directly to compliance-officer@secure-data.io ✉️, or you can leave a secure message on their dedicated hotline at +1-800-555-0199. 🔒
"""
```

```python
emojis = [em for em in emoji_text if emoji.is_emoji(em)]
```

```python
print(emojis)
```

```python
text_without_emoji = emoji.demojize(emoji_text, delimiters=("", " "))
text_without_emoji
```

# Punctuations

```python
import string
splitted_para = text_without_emoji.split()
just_words = [char for char in splitted_para if char not in string.punctuation]
word_count = len(just_words)
print(word_count)
```

```python
unique_words = set(just_words)
print(f"Found {len(unique_words)} unique words: \n")
for i in unique_words:
    print(i)
```
